```python
import pandas as pd

df = pd.read_csv(r"D:\OMKAR\Python\Job Career Analysis\indian_tech_career_intelligence_2026.csv")

df
```

```python
pd.set_option('display.max.rows', 1000)
```

```python
df
```

```python
import numpy as np

df_clean = df.copy()

# Normalize column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
df
```

```python
df_clean = df_clean.drop_duplicates().copy()

# Convert dates
for col in ["posted_date", "application_deadline"]:
    df_clean[col] = pd.to_datetime(
        df_clean[col], errors="coerce"
    )

# Convert numeric columns
numeric_cols = [
    "salary_minimum",
    "salary_maximum",
     "stipend_amount",
    "skill_count",
    "experience_required",
    "applicant_count",
    "career_growth_index",
]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors="coerce"
    )

# Inspect cleaning results
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Missing values:")
print(df_clean.isna().sum().sort_values(ascending=False).head(15))
```

```python
# Top 10 job categories
category_demand = (
    df_clean["job_category"]
    .value_counts()
    .head(10)
)

display(category_demand)

# Top cities
city_demand = (
    df_clean["location_city"]
    .value_counts()
    .head(10)
)

display(city_demand)

# Work mode distribution
display(df_clean["work_mode"].value_counts())

# Experience level distribution
display(df_clean["experience_level"].value_counts())
```

```python
salary_df = df_clean[
    (df_clean["salary_minimum"] > 0) &
    (df_clean["salary_maximum"] >= df_clean["salary_minimum"])
].copy()

salary_by_role = (
    salary_df.groupby("job_category")
    .agg(
        job_count=("job_title", "count"),
        median_min_salary=("salary_minimum", "median"),
        median_max_salary=("salary_maximum", "median")
    )
    .sort_values("median_max_salary", ascending=False)
)

display(salary_by_role.head(15))
```

```python
skill_series = (
    df_clean["skills_extracted"]
    .dropna()
    .astype(str)
    .str.split(",")
    .explode()
    .str.strip()
    .str.lower()
)

skill_counts = (
    skill_series[skill_series.ne("")]
    .value_counts()
    .head(20)
)

display(skill_counts)
```
